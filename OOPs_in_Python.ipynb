# **ATM Class**

```python
class Atm:
  def __init__(self):
    self.pin = ""      # instance variable
    self.balance = 0   # instance variable

    print(id(self)) # to justify that self is the current object itself by checking add of self and the current object.
    print("Hello, Welcome to the ATM")
    self.menu()

  def menu(self):
    user_input = input("""
    How, would you like to proceed?
    1. Enter 1 to create pin
    2. Enter 2 to deposit
    3. Enter 3 to withdraw
    4. Enter 4 to check balance
    5. Enter 5 to exit
    """)

    if user_input == "1":
      print("Create Pin")
      self.create_pin()
    elif user_input == "2":
      print("Deposit")
      self.deposit()
    elif user_input == "3":
      print("Withdraw")
      self.withdraw()
    elif user_input == "4":
      print("Check Balance")
      self.check_balance()
    else:
      print("Exit")

  def create_pin(self):
    self.pin = input("Enter your 4 digit pin")
    print("Pin set successfully")


  def deposit(self):
    temp_pin = input("Enter your 4 digit pin to proceed\n")
    if temp_pin == self.pin:
      deposit_amount = float(input("Enter the amount to deposit\n"))
      self.balance+=deposit_amount
      print(f"{deposit_amount} deposited successfully.")
    else:
      print("Wrong PIN, Please try again!")

  def withdraw(self):
    temp_pin = input("Enter your 4 digit pin to proceed\n")
    if temp_pin == self.pin:
      withdraw_amount = float(input("Enter the amount to withdraw\n"))
      if withdraw_amount <= self.balance:
        self.balance-=withdraw_amount
        print(f"{withdraw_amount} withdrawn successfully.")
      else:
        print("Insufficient funds")
    else:
      print("Wrong PIN, Please try again!")

  def check_balance(self):
    temp_pin = input("Enter your 4 digit pin to proceed\n")
    if temp_pin == self.pin:
      print(f"Your balance is {self.balance}")
    else:
      print("Wrong PIN, Please try again!")
```

```python
class Fraction:

  def __init__(self, n, d): #constructor
    self.num = n
    self.den = d

  def __str__(self): #dunder method __str__ -> automatically called when the object of this class is written inside print function
    # print("Hello, creating a new datatype to handle fractional values in Python\n")

    return f"{self.num}/{self.den}"

    # print(f"{self.num}/{self.den}")


# the __str__ method must return a string type, by using print statement in __str__ (return type is NONE) so it gives an error
# if we change the name of __str__ to __strx__ or anything else, there will be no error but it'll not behave like a magic method now,
# which means this method will not automatically run when we put the object of the Fraction class i print statement

  # method to add two fractions in Python
  def __add__(self, other):  # other is the second fraction
    tempp_num = self.num*other.den + self.den*other.num
    tempp_den = self.den * other.den
    return "{}/{}".format(tempp_num, tempp_den)

  # method to subtract 2 fractions:
  def __sub__(self, other):
    temp_num = self.num*other.den - self.den*other.num
    temp_den = self.den * other.den
    return f"{temp_num}/{temp_den}"

  # method to multiple 2 fraction:
  def __mul__(self, other):
    temp_num = self.num * other.num
    temp_den = self.den * other.den
    return f"{temp_num}/{temp_den}"

  # method to divide 2 fraction:
  def __truediv__(self, other):
    temp_num = self.num * other.den
    temp_den = self.den * other.num
    return f"{temp_num}/{temp_den}"
```

# **Operator Overloading in the above Fraction class code:**

When we “overload” an operator, we are telling Python how to interpret that operator when used with our own custom objects.

Normally + adds numbers (int.__add__),
But we decided that for Fraction, + should add two fractions.

Basically, we are giving new meaning to the existing + operator, specifically for our class.

That’s why it’s called operator overloading:
We’re overloading (redefining) the meaning of the operator

Based on the type of operands involved.

```python
# Another sample Code to understand Operator Overloading in Python OOPs
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks

s1 = Student("Akshat", 85)
s2 = Student("Riya", 90)

print(s1 + s2)   # Output: 175

# Here, the + operator calls the magic method __add__(). So we’ve overloaded the + operator to add marks instead of normal numbers.
# Operator overloading is basically redefining the operators for custom objects.
```

# **Let's see another simple code of Magic/ Dunder methods in Python OOPs**

## __str__ → prints readable string
## __len__ → lets len() work
## __add__ → lets + work on custom objects

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"{self.title} ({self.pages} pages)"

    def __len__(self):
        return self.pages

    def __add__(self, other):
        return self.pages + other.pages

book1 = Book("Python Basics", 200)
book2 = Book("Data Science", 300)

print(book1)        # Python Basics (200 pages)
print(len(book1))   # 200
print(book1 + book2)  # 500
```

```python
# Instance Variables:

class Car:
    def __init__(self, brand, color):
        self.brand = brand        # instance variable
        self.color = color        # instance variable
        print(id(self))

c1 = Car("Tesla", "Red")
c2 = Car("BMW", "Black")

print(c1.brand)   # Tesla
print(c2.brand)   # BMW

# print(self.brand)

# this will give an error as self is not a keyword in Python.
# self should be used inside a class as the current object but outside the class Python does not know self and hence an error

# Each car object has its own copy of brand and color.
# Changing c1.color won’t affect c2.color.
# Instance variables = object’s personal data.
```

```python
# Class Variables:

class Car:
    wheels = 4   # class variable (common for all)

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

c1 = Car("Tesla", "Red")
c2 = Car("BMW", "Black")

print(c1.wheels)  # 4
print(c2.wheels)  # 4

Car.wheels = 6
print(c1.wheels)  # 6
print(c2.wheels)  # 6


# Class variables are shared memory across all objects.

# If you change it through the class name, the change is visible everywhere.

# Example in real life: “All cars have 4 wheels” → this property belongs to the Car class, not each specific car.
```

```python
# Static Variables:

class MathTools:
    PI = 3.14159  # static/class variable

    @staticmethod
    def circle_area(radius):
        return MathTools.PI * radius ** 2


# PI is a static/class variable
# It’s used by the class, not by individual instances.
```

```python
class Employee:
    company_name = "TechCorp"  # class variable

    def __init__(self, name, salary):
        self.name = name       # instance variable
        self.salary = salary   # instance variable

    @staticmethod
    def company_policy():
        return "Work from home allowed 2 days/week"  # static method

e1 = Employee("Akshat", 70000)
e2 = Employee("Bharti", 80000)

print(e1.company_name)   # TechCorp as it is class variable (same for all objects)
print(e2.company_name)   # TechCorp as it is class variable (same for all objects)
print(e1.name)           # Akshat as it is instance variable (diff for all objects)
print(e2.name)           # Riya as it is instance variable (diff for all objects)

Employee.company_name = "DataWorks" # class variable updated
print(e1.company_name)   # DataWorks (now we see the updated value)
print(e2.company_name)   # DataWorks (now we see the updated value)

# company_policy() → static method (belongs to the class)
```

# **Note**:
Encapsulation isn’t meant to make it impossible to access data — it’s meant to discourage you from doing it.

The __ double underscore is a gentle way of saying:

“This variable is for internal use only — please don’t touch it from outside.”

You can still access it (via name mangling), but you shouldn’t.
That’s why Python allows it but makes it inconvenient — to maintain flexibility and responsibility.

# **We can use getter and setter methods to access and update the values of private variables**

# **Example 2: Encapsulation**


```python
class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self._dept = "Finance"    # protected
        self.__salary = salary    # private

    def show_info(self):
        print(f"Employee: {self.name}, Dept: {self._dept}")

    def get_salary(self):  #getter method to get the value of private variables
        return self.__salary

    def set_salary(self, amount): # setter method to set/ update the values of private variables
        if amount > 0:
            self.__salary = amount
        else:
            print("Invalid salary")

emp = Employee("Akshat", 50000)

print(emp.name)          # Accessible
print(emp._dept)         # ⚠️ Accessible, but not recommended
print(emp.get_salary())  # Access via method
emp.set_salary(60000)
print(emp.get_salary())  # Updated salary


# __salary is private → can’t access directly like emp.__salary.
# We use getter and setter methods (get_salary, set_salary) for controlled access.
# That’s the core of encapsulation — we’re hiding data and exposing it safely.
```

# **Example 3: Encapsulation**

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance   # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit must be positive!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds or invalid amount!")

    def get_balance(self):
        return self.__balance

# Usage
account = BankAccount("Riya", 1000)
account.deposit(500)
print(account.get_balance())  #1500
account.withdraw(2000)        #Insufficient funds

# Here, balance is hidden to prevent misuse (like account.__balance = -500).
```

```python
class BankAccount:
    def __init__(self, owner, balance, password):
        # Public attribute
        self.owner = owner

        # Private attributes (encapsulated data)
        self.__balance = balance
        self.__password = password

    # Getter Method (takes password as ARGUMENT)
    # Useful in automated systems where password can be passed programmatically
    def get_balance(self, password):
        if password == self.__password:
            print(f"Account Holder: {self.owner}")
            return f"Your current balance is ₹{self.__balance}"
        else:
            return "Access Denied: Incorrect password."

    # Setter Method (asks password INSIDE method)
    # Useful in user-facing applications (like an ATM)
    def set_balance(self):
        entered_password = input("Enter your password to update balance: ")
        if entered_password == self.__password:
            new_balance = float(input("Enter the new balance amount: "))
            if new_balance >= 0:
                self.__balance = new_balance
                print("Balance updated successfully!")
            else:
                print("Invalid balance amount. Must be non-negative.")
        else:
            print("Access Denied: Incorrect password.")

    # Regular methods for deposit/withdraw operations
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit must be positive!")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")
        else:
            print("Insufficient funds or invalid amount!")


account = BankAccount("Riya", 1000, "riya123")

# Deposit money
account.deposit(500)

# Getter method → Password passed as ARGUMENT
print(account.get_balance("riya123"))   # Correct password
print(account.get_balance("wrong123"))  # Incorrect password

# Setter method → Password taken INSIDE the method (interactive)
account.set_balance()

# Trying to directly access private variable (will fail)
# print(account.__balance)  # AttributeError: private variable
```

## **Types of Inheritance in Python**

```python
# Single Inheritance: One parent → One child.

class Parent:
    def greet(self):
        print("Hello from Parent")

class Child(Parent):
    def message(self):
        print("Hello from Child")

c = Child()
c.greet() # greet is a method of parent classs but chile can access it using inheritance
c.message()
```

```python
# Multiple Inheritance: One child → Multiple parents.

class Father:
    def skills(self):
        print("I can drive")

class Mother:
    def skills(self):
        print("I can cook")

class Child(Father, Mother):
    def more_skills(self):
        print("I can code")

c = Child()
c.skills()        # Python follows Method Resolution Order (MRO)
c.more_skills()

# Python will call the first parent’s method in case of conflict (Father first in this case).
# As above we have skills() method in both parent classes. So, the child will access skills() of Father class before Mother class
```

```python
# Multilevel Inheritance: Parent → Child → Grandchild.

class Grandfather:
    def show(self):
        print("I am the Grandfather")

class Father(Grandfather):
    def show_father(self):
        print("I am the Father")

class Son(Father):
    def show_son(self):
        print("I am the Son")

s = Son()
s.show()
s.show_father()
s.show_son()

#s.x()

# The Son indirectly inherits from both Father and Grandfather.
```

```python
# Hierarchical Inheritance: One parent → Multiple children.

class Parent:
    def show(self):
        print("Parent class method")

class Child1(Parent):
    def child1_info(self):
        print("Child 1 method")

class Child2(Parent):
    def child2_info(self):
        print("Child 2 method")

obj1 = Child1()
obj1.show()
obj1.child1_info()

obj2 = Child2()
obj2.show()
obj2.child2_info()


# Both children share the same parent class.
```

```python
# Hybrid Inheritance: Combination of multiple and multilevel inheritance.

class A:
    def featureA(self):
        print("Feature A")

class B(A):
    def featureB(self):
        print("Feature B")

class C(A):
    def featureC(self):
        print("Feature C")

class D(B, C):  # Hybrid
    def featureD(self):
        print("Feature D")

d = D()
d.featureA()
d.featureB()
d.featureC()
d.featureD()


#It’s a mix, but in complex hierarchies, Python handles conflicts via MRO (Method Resolution Order).
```

# **MRO in Inheritance**

Imagine you are raising a customer service issue.

D (front desk) receives it first, handles its part, then passes it to

The order of escalation is your MRO.

```python
# If a method or variable name appears in multiple parent classes, MRO decides which one gets picked first.”

# MRO (Method Resolution Order) defines the order in which Python looks for a method or attribute when it is called on an object —
# particularly when multiple classes are involved in inheritance.

class A:
    def show(self):
        print("Class A")

class B(A):
    def show(self):
        print("Class B")

class C(A):
    def show(self):
        print("Class C")

class D(B, C):
    print("Class D")


obj = D()
obj.show()

print(D.mro())   # or D.__mro__

# Explanation

# D inherits from both B and C. So Python must decide: when obj.show() is called, should it run show() from B or C?

# MRO defines the order:

# D → B → C → A → object

# So when obj.show() is called: Python first checks in D → not found. Then in B → found there. So it executes Class B.
```

```python
# Example 2: Using super() with MRO: Let’s see how super() follows the same MRO chain 👇

class A:
    def show(self):
        print("A class method")

class B(A):
    def show(self):
        print("B class method")
        super().show()

class C(A):
    def show(self):
        print("C class method")
        super().show()

class D(B, C):
    def show(self):
        print("D class method")
        super().show()

d = D()
d.show()

print(D.mro())

# Explanation

# Each show() calls super().show(), which doesn’t mean “parent,” but “next class in MRO.”

# So the chain of calls follows:
```

# **Method Overriding vs Method Overloading vs Operator Overloading**


# 1. Method Overriding (OOP concept)

Definition:
 *When a subclass defines a method with the same name as a method in its parent class, and the subclass version replaces (or overrides) the parent version.*

In short —
 Same method name
 Same parameters
 Different behavior (in child class)

Use case:
We override methods to provide specialized behavior for subclasses while keeping a consistent interface.



# 2. Method Overloading

Definition:
 *When multiple methods have the same name but different parameters or argument types.*

In many languages like Java or C++, this is allowed (compile-time polymorphism).
But in Python, true method overloading doesn’t exist — because the last defined method with the same name overwrites the previous ones.

However, Python can simulate overloading using:

* Default arguments
* Variable-length arguments (*args, kwargs)

# 3. Operator Overloading

Definition:
 *It’s a special type of method overloading where operators like +, -, >, etc. are made to work with user-defined objects.*

This uses magic (dunder) methods like __add__, __sub__, etc.

```python
# for mutable types like list, set, dictionary, etc

def modify_list(lst):
    print("The address of lst inside function before append ", id(lst))
    lst.append(100)
    print("The address of lst inside function after append ", id(lst))
    print("Inside function:", lst)

nums = [1, 2, 3]
print("The address of nums before function call ", id(nums))
modify_list(nums) # nums is passed as refernce to the list object with value [1,2,3]
# num will be assigned to lst (both pointing the same location)

print("The address of nums before function call ", id(nums))
print("Outside function:", nums)
```

```python
# pass by reference for immutable types


def change_number(n):
    print("The address of n inside function before update",id(n))
    n = n + 10
    print("The address of n inside function after update",id(n))
    print("Inside function:", n)

x = 5
print("The address of x is before function call",id(x))
change_number(x)
print("The address of x is after function call",id(x))
print("Outside function:", x)

# integers are immutable, so changing n inside the function creates a new object, leaving the original x unchanged.

#Explanation:

# When you call change_number(x),
# → The value of x (5) is passed by reference to the parameter n.

# Initially, both x and n refer to the same object (5) in memory. So, same address of x and n

# When you do n = n + 10,
# → Python creates a new integer object (15), because integers are immutable. So, the address of n changes.
# → Now n points to a new address, but x still points to the old one (5).

# After the function ends, n disappears — it was a local variable, and x remains unchanged.

```

# **Must understand: Pass by reference vs pass by Assignment froma above code**

In Python:

Everything is an object, and variables hold references to those objects.

But whether a variable shares or copies an object’s behavior depends on the object’s type:

Immutable types (int, str, tuple): A new object is created when you “modify” it.

Mutable types (list, dict, set): The same object is modified in place.

So:

When you do b = a, b gets a reference to the same object as a.

When you do a = 7, you’re not changing the object — you’re changing which object a refers to.

That’s why it’s not pass-by-reference — it’s “pass-by-object-reference” or “pass-by-assignment”, meaning:

Variables are names (references) bound to objects, and reassignment changes what the name points to — not the object itself.

# **Abstraction in Python OOPs**

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car engine started...")

    def stop(self):
        print("Car engine stopped...")

# v = Vehicle()  # ❌ Error: can't instantiate abstract class
c = Car()
c.start()
c.stop()
```

```python
# Real-World Banking System with Abstraction

from abc import ABC, abstractmethod

class BankAccount(ABC):
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def withdraw(self, amount):
        pass

class SavingsAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ₹{amount}, New Balance: ₹{self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrawn ₹{amount}, Remaining Balance: ₹{self.balance}")
        else:
            print("Insufficient Balance")

class CurrentAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ₹{amount}, New Balance: ₹{self.balance}")

    def withdraw(self, amount):
        if amount <= (self.balance + 10000):  # Overdraft limit
            self.balance -= amount
            print(f"Withdrawn ₹{amount}, Remaining Balance: ₹{self.balance}")
        else:
            print("Overdraft Limit Reached!")

# Abstraction in action
accounts = [SavingsAccount("SA101"), CurrentAccount("CA202")]

for acc in accounts:
    acc.deposit(5000)
    acc.withdraw(2000)
```
